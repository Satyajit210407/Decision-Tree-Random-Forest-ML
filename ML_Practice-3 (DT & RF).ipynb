## **Part A: Iris Dataset**

Target variable: `species` (3 classes)

**Q1.** Import the required libraries and load the `iris` dataset from Seaborn. Display the first 5 rows.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

data = sns.load_dataset('iris')
data.head()
```

**Q2.** Check the shape of the dataset, data types, and confirm there are no missing values.

```python
data.info()
```

**Q3.** Define the independent variables `X` (all feature columns) and the dependent variable `y` (`species`).

```python
X_iris = data.drop('species', axis=1)
y_iris = data['species']
```

**Q4.** Split the dataset into training and testing sets.

```python
from sklearn.model_selection import train_test_split
X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(X_iris, y_iris, test_size=0.2, random_state=42)
```

```python
X_train_iris
```

```python
y_train_iris
```

### Decision Tree Classifier — Iris

**Q5.** Build a Decision Tree Classifier and train it on the training data.

```python
from sklearn.tree import DecisionTreeClassifier
classifier= DecisionTreeClassifier(criterion='entropy', random_state=0)
classifier.fit(X_train_iris, y_train_iris)
```

**Q6.** Predict the `species` for the test set.

```python
y_pred_iris= classifier.predict(X_test_iris)
```

**Q7.** Generate and print the Classification Report for the Decision Tree model.

```python
from sklearn.metrics import classification_report

print("Classification Report for Decision Tree Model:")
print(classification_report(y_test_iris, y_pred_iris))
```

**Q8.** Generate and print the Confusion Matrix for the Decision Tree model.

```python
from sklearn.metrics import confusion_matrix
dt_cm_iris = confusion_matrix(y_test_iris, y_pred_iris)
print("Confusion Matrix for Decision Tree Model:")
print(dt_cm_iris)
```

**Q9.** Calculate and print the Accuracy Score for the Decision Tree model.

```python
from sklearn.metrics import accuracy_score
dt_accuracy_iris = accuracy_score(y_test_iris, y_pred_iris)
print("Accuracy of Decision Tree Model:", dt_accuracy_iris)
```

### Random Forest Classifier — Iris

**Q10.** Build a Random Forest Classifier and train it on the training data.

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train_iris, y_train_iris)
```

**Q11.** Predict the `species` for the test set.

```python
y_pred_rf_iris = rf.predict(X_test_iris)
```

**Q12.** Generate and print the Classification Report for the Random Forest model.

```python
from sklearn.metrics import classification_report
print("Classification Report for Random Forest Model:")
print(classification_report(y_test_iris, y_pred_rf_iris))
```

**Q13.** Generate and print the Confusion Matrix for the Random Forest model.

```python
from sklearn.metrics import confusion_matrix
rf_cm_iris = confusion_matrix(y_test_iris, y_pred_rf_iris)
print("Confusion Matrix for Random Forest Model:")
print(rf_cm_iris)
```

**Q14.** Calculate and print the Accuracy Score for the Random Forest model.

```python
from sklearn.metrics import accuracy_score
rf_accuracy_iris = accuracy_score(y_test_iris, y_pred_rf_iris)
print("Accuracy of Random Forest Model:", rf_accuracy_iris)
```

**Q15.** Compare the accuracy scores of the Decision Tree and Random Forest models on the Iris dataset. Which model performed better, and by how much?

```python
difference = abs(rf_accuracy_iris - dt_accuracy_iris) * 100

if dt_accuracy_iris > rf_accuracy_iris:
    print(f"Decision Tree performed better by {difference:.2f}%")
elif rf_accuracy_iris > dt_accuracy_iris:
    print(f"Random Forest performed better by {difference:.2f}%")
else:
    print("Both models performed equally.")
```

## **Part B: Tips Dataset**

Target variable: `time` (2 classes — Lunch/Dinner)

**Q16.** Load the `tips` dataset from Seaborn. Display the first 5 rows.

```python
data_tips = sns.load_dataset('tips')
data_tips.head()
```

**Q17.** Check the shape of the dataset, data types, and confirm there are no missing values.

```python
data_tips.info()
```

**Q18.** Identify the categorical columns in the dataset and encode them into numeric form so they can be used as model inputs.

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data_tips['sex_encoded'] = le.fit_transform(data_tips['sex'])
data_tips['smoker_encoded'] = le.fit_transform(data_tips['smoker'])
data_tips['day_encoded'] = le.fit_transform(data_tips['day'])
data_tips['time_encoded'] = le.fit_transform(data_tips['time'])

data_tips
```

**Q19.** Define the independent variables `X` (all relevant feature columns) and the dependent variable `y` (`time`).

```python
X_tips = data_tips[['total_bill', 'tip', 'sex_encoded','smoker_encoded', 'day_encoded', 'size']]
y_tips = data_tips['time_encoded']
```

**Q20.** Split the dataset into training and testing sets.

```python
from sklearn.model_selection import train_test_split
X_train_tips, X_test_tips, y_train_tips, y_test_tips = train_test_split(X_tips, y_tips, test_size=0.2, random_state=42)
```

```python
X_train_tips
```

```python
y_train_tips
```

### Decision Tree Classifier — Tips

**Q21.** Build a Decision Tree Classifier and train it on the training data.

```python
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier(criterion='entropy', random_state=42)
clf.fit(X_train_tips, y_train_tips)
```

**Q22.** Predict `time` for the test set.

```python
y_pred_tips = clf.predict(X_test_tips)
```

**Q23.** Generate and print the Classification Report for the Decision Tree model.

```python
from sklearn.metrics import classification_report
print("Classification Report for Decision Tree Model on Tips Dataset:")
print(classification_report(y_test_tips, y_pred_tips))
```

**Q24.** Generate and print the Confusion Matrix for the Decision Tree model.

```python
from sklearn.metrics import confusion_matrix
dt_cm_tips = confusion_matrix(y_test_tips, y_pred_tips)
print("Confusion Matrix for Decision Tree Model on Tips Dataset:")
print(dt_cm_tips)
```

**Q25.** Calculate and print the Accuracy Score for the Decision Tree model.

```python
from sklearn.metrics import accuracy_score
dt_accuracy_tips = accuracy_score(y_test_tips, y_pred_tips)
print("Accuracy of Decision Tree Model on Tips Dataset:", dt_accuracy_tips)
```

### Random Forest Classifier — Tips

**Q26.** Build a Random Forest Classifier and train it on the training data.

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(criterion='entropy',random_state=42)
rf_model.fit(X_train_tips, y_train_tips)
```

**Q27.** Predict `time` for the test set.

```python
y_pred_rf_tips = rf_model.predict(X_test_tips)
```

**Q28.** Generate and print the Classification Report for the Random Forest model.

```python
from sklearn.metrics import classification_report

print("Classification Report for Random Forest Model on Tips Dataset:")
print(classification_report(y_test_tips, y_pred_rf_tips))
```

**Q29.** Generate and print the Confusion Matrix for the Random Forest model.

```python
from sklearn.metrics import confusion_matrix
rf_cm_tips = confusion_matrix(y_test_tips, y_pred_rf_tips)
print("Confusion Matrix for Random Forest Model on Tips Dataset:")
print(rf_cm_tips)
```

**Q30.** Calculate and print the Accuracy Score for the Random Forest model.

```python
from sklearn.metrics import accuracy_score
rf_accuracy_tips = accuracy_score(y_test_tips, y_pred_rf_tips)
print("Accuracy of Random Forest Model on Tips Dataset:", rf_accuracy_tips)
```

**Q31.** Compare the accuracy scores of the Decision Tree and Random Forest models on the Tips dataset. Which model performed better, and by how much?

```python
difference_tips = abs(rf_accuracy_tips - dt_accuracy_tips) * 100

if dt_accuracy_tips > rf_accuracy_tips:
    print(f"Decision Tree performed better on Tips Dataset by {difference_tips:.2f}%")

elif rf_accuracy_tips > dt_accuracy_tips:
    print(f"Random Forest performed better on Tips Dataset by {difference_tips:.2f}%")

else:
    print("Both models performed equally on Tips Dataset.")
```

## **Final Comparison**

**Q32.** Summarize the accuracy scores of all four models (Decision Tree – Iris, Random Forest – Iris, Decision Tree – Tips, Random Forest – Tips) in a single table or DataFrame. Which dataset-model combination performed best overall?

```python
accuracy_df = pd.DataFrame({
    'Dataset': ['Iris', 'Iris', 'Tips', 'Tips'],
    'Model': [
        'Decision Tree',
        'Random Forest',
        'Decision Tree',
        'Random Forest'
    ],
    'Accuracy': [
        dt_accuracy_iris,
        rf_accuracy_iris,
        dt_accuracy_tips,
        rf_accuracy_tips
    ]
})

accuracy_df['Accuracy (%)'] = accuracy_df['Accuracy'] * 100
print(accuracy_df)

best_model = accuracy_df.iloc[0]

print("\nBest-performing combination overall:")
print(f"Dataset: {best_model['Dataset']}")
print(f"Model: {best_model['Model']}")
print(f"Accuracy: {best_model['Accuracy (%)']:.2f}%")
```
